## Question 1: Using Classes and OOP

Make a class called `Employee`. Its `__init__()` method should store the employee's name, number, address, salary, post, year of joining, education, and years of experience.

Add methods to:
- Display the employee's information (`describe_employee()`).
- Update years of experience (`update_year_of_experience()`).
- Increase the employee's salary (`increase_salary()`).

Create an instance named `employee101`.


```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, years_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.years_of_experience = years_of_experience

    def describe_employee(self):
        """Print all stored employee details."""
        print("Employee Details")
        print("----------------")
        print(f"Name: {self.name}")
        print(f"Employee Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: {self.salary:,.2f}")
        print(f"Post: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Years of Experience: {self.years_of_experience}")

    def update_year_of_experience(self, new_experience):
        """Update the employee's total years of experience."""
        if new_experience < 0:
            raise ValueError("Years of experience cannot be negative.")
        self.years_of_experience = new_experience
        print(f"Experience updated to {self.years_of_experience} years.")

    def increase_salary(self, amount):
        """Increase salary by a fixed amount."""
        if amount < 0:
            raise ValueError("Salary increase cannot be negative.")
        self.salary += amount
        print(f"Salary increased by {amount:,.2f}. New salary: {self.salary:,.2f}")


# Create the required employee101 instance.
employee101 = Employee(
    name="Ali Ahmed",
    number=101,
    address="Karachi, Pakistan",
    salary=85000,
    post="Sales Executive",
    year_of_joining=2024,
    education="Bachelor's Degree",
    years_of_experience=2
)

# Display the employee's details.
employee101.describe_employee()

# Example method calls (uncomment to test updates):
# employee101.update_year_of_experience(3)
# employee101.increase_salary(5000)
```

## Question 2: Complete the same task without Classes or OOP

Use a dictionary to store the employee details and ordinary functions to display and update them. This performs the same operations as Question 1 without defining a class.


```python
# Store employee information in a dictionary instead of an object.
employee101_data = {
    "name": "Ali Ahmed",
    "number": 101,
    "address": "Karachi, Pakistan",
    "salary": 85000,
    "post": "Sales Executive",
    "year_of_joining": 2024,
    "education": "Bachelor's Degree",
    "years_of_experience": 2
}


def describe_employee(employee):
    """Print all employee details."""
    print("Employee Details")
    print("----------------")
    print(f"Name: {employee['name']}")
    print(f"Employee Number: {employee['number']}")
    print(f"Address: {employee['address']}")
    print(f"Salary: {employee['salary']:,.2f}")
    print(f"Post: {employee['post']}")
    print(f"Year of Joining: {employee['year_of_joining']}")
    print(f"Education: {employee['education']}")
    print(f"Years of Experience: {employee['years_of_experience']}")


def update_year_of_experience(employee, new_experience):
    """Update years of experience in the dictionary."""
    if new_experience < 0:
        raise ValueError("Years of experience cannot be negative.")
    employee["years_of_experience"] = new_experience
    print(f"Experience updated to {employee['years_of_experience']} years.")


def increase_salary(employee, amount):
    """Increase salary by a fixed amount."""
    if amount < 0:
        raise ValueError("Salary increase cannot be negative.")
    employee["salary"] += amount
    print(f"Salary increased by {amount:,.2f}. New salary: {employee['salary']:,.2f}")


# Display the employee's details.
describe_employee(employee101_data)

# Example function calls (uncomment to test updates):
# update_year_of_experience(employee101_data, 3)
# increase_salary(employee101_data, 5000)
```
